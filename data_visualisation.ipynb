```python
import json
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# -----------------------------
# Load JSON glauber_cooling
# -----------------------------
json_file_glauber_cooling = r"C:\Users\siauw\OneDrive\Desktop\Modelling and Visualisation\Ising Model\glauber_50_cooling\Observables.json"
json_file_kawasaki_cooling = r"C:\Users\siauw\OneDrive\Desktop\Modelling and Visualisation\Ising Model\kawasaki_50_cooling\Observables.json"
json_file_glauber_heating = r"C:\Users\siauw\OneDrive\Desktop\Modelling and Visualisation\Ising Model\glauber_50_heating\Observables.json"
json_file_kawasaki_heating = r"C:\Users\siauw\OneDrive\Desktop\Modelling and Visualisation\Ising Model\kawasaki_50_heating\Observables.json"

with open(json_file_glauber_cooling, "r") as f:
    glauber_cooling = json.load(f)

with open(json_file_kawasaki_cooling, "r") as f:
    kawasaki_cooling = json.load(f)

with open(json_file_glauber_heating, "r") as f:
    glauber_heating = json.load(f)

with open(json_file_kawasaki_heating, "r") as f:
    kawasaki_heating = json.load(f)
```

```python
def plot_E(glauber, kawasaki):
    # Glauber table
    glauber_df = pd.DataFrame({
        "T": np.array(glauber["T_vals"]),
        "Energy": np.array(glauber["energy"]["values"]),
        "Energy_err": np.array(glauber["energy"]["errors"]),
        "Cv": np.array(glauber["heat_capacity"]["values"]),
        "Cv_err": np.array(glauber["heat_capacity"]["errors"])
    })
    
    # Kawasaki table
    kawasaki_df = pd.DataFrame({
        "T": np.array(kawasaki["T_vals"]),
        "Energy": np.array(kawasaki["energy"]["values"]),
        "Energy_err": np.array(kawasaki["energy"]["errors"]),
        "Cv": np.array(kawasaki["heat_capacity"]["values"]),
        "Cv_err": np.array(kawasaki["heat_capacity"]["errors"])
    })
    
    fig1, axes1 = plt.subplots(1, 2, figsize=(15, 8), sharex=True)

    # Subplot 1: Average Energy
    axes1[0].errorbar(glauber_df["T"], glauber_df["Energy"], yerr=glauber_df["Energy_err"], fmt='o-', markersize=4, elinewidth=1.5, capsize=3, color='tab:blue', label='Glauber')
    axes1[0].errorbar(kawasaki_df["T"], kawasaki_df["Energy"], yerr=kawasaki_df["Energy_err"], fmt='o-', markersize=4, elinewidth=1.5, capsize=3, color='tab:orange', label='Kawasaki')
    axes1[0].set_ylabel(r"Average Energy  $\langle E \rangle$ [J]")
    axes1[0].set_xlabel(r"Temperature [$k_bT$]")
    axes1[0].set_title(r"Average Energy $\langle E \rangle$ against Temperature [$k_bT$]")
    axes1[0].legend()

    # Subplot 2: Heat Capacity
    axes1[1].errorbar(glauber_df["T"], glauber_df["Cv"], yerr=glauber_df["Cv_err"], fmt='s-', markersize=4, elinewidth=1.5, capsize=3, color='tab:blue', label='Glauber')
    axes1[1].errorbar(kawasaki_df["T"], kawasaki_df["Cv"], yerr=kawasaki_df["Cv_err"], fmt='s-', markersize=4, elinewidth=1.5, capsize=3, color='tab:orange', label='Kawasaki')
    axes1[1].set_xlabel(r"Temperature [$k_bT$]")
    axes1[1].set_ylabel(r"Heat Capacity $C_v$ [$k_b$]")
    axes1[1].set_title(r"Heat Capacity [$C_v$] against Temperature [$k_bT$]")
    axes1[1].legend()

    plt.tight_layout()
    plt.show()

    print("Glauber")
    display(glauber_df)
    
    print("Kawasaki")
    display(kawasaki_df)

    # return glauber_df, kawasaki_df


def plot_M(glauber):
    # Glauber table
    glauber_df = pd.DataFrame({
        "T": np.array(glauber["T_vals"]),
        "Magnetisation": np.array(glauber["magnetization"]["values"]),
        "Magnetisation_err": np.array(glauber["magnetization"]["errors"]),
        "Susceptibility": np.array(glauber["susceptibility"]["values"]),
        "Susceptibility_err": np.array(glauber["susceptibility"]["errors"])
    })

    fig2, axes2 = plt.subplots(1, 2, figsize=(15, 8), sharex=True)

    # Subplot 1: Average Magnetization
    axes2[0].errorbar(glauber_df["T"], glauber_df["Magnetisation"], yerr=glauber_df["Magnetisation_err"], fmt='o-', markersize=4, elinewidth=1.5, capsize=3, color='tab:blue')
    axes2[0].set_ylabel(r"Average Magnetization $\langle |M| \rangle$")
    axes2[0].set_xlabel(r"Temperature [$k_bT$]")
    axes2[0].set_title(r"Average Magnetization $\langle |M| \rangle$ against Temperature [$k_bT$]")

    # Subplot 2: Susceptibility
    axes2[1].errorbar(glauber_df["T"], glauber_df["Susceptibility"], yerr=glauber_df["Susceptibility_err"], fmt='s-', markersize=4, elinewidth=1.5, capsize=3, color='tab:blue')
    axes2[1].set_xlabel(r"Temperature [$k_bT$]")
    axes2[1].set_ylabel(r"Susceptibility $\chi$")
    axes2[1].set_title(r"Magnetic Susceptibility [$\chi$] against Temperature [$k_bT$]")

    plt.tight_layout()
    plt.show()

    print("Glauber")
    display(glauber_df)

    # return glauber_df
```

```python
plot_E(glauber_cooling, kawasaki_cooling)
```

```python
plot_M(glauber_cooling)
```

```python
plot_E(glauber_heating, kawasaki_heating)
```

```python
plot_M(glauber_heating)
```
